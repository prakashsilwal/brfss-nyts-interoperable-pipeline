```python
from pyspark.sql import functions as F

def safe_int(col):
    """Cast to int, treating empty/blank strings as null instead of erroring."""
    return F.when((F.col(col) == "") | F.col(col).isNull(), None).otherwise(F.col(col).cast("int"))

cols_needed = [
    "_TOTINDA", "GENHLTH", "PHYSHLTH", "MENTHLTH", "_BMI5", "SMOKE100",
    "EDUCA", "INCOME3", "EMPLOY1",
    "DIABETE4", "CVDINFR4", "CVDSTRK3", "HAVARTH4", "ADDEPEV3",
    "SEXVAR", "_AGEG5YR", "_RACE",
]

df = bronze_df.select(*cols_needed)

# Apply safe_int to every column that needs numeric comparisons
for c in cols_needed:
    df = df.withColumn(c, safe_int(c))

# Recode special codes to NaN per Section 3.4 of the manuscript
df = df.withColumn("_TOTINDA", F.when(F.col("_TOTINDA") == 9, None).otherwise(F.col("_TOTINDA")))

binary_disease_cols = ["DIABETE4", "CVDINFR4", "CVDSTRK3", "HAVARTH4", "ADDEPEV3"]
for c in binary_disease_cols:
    df = df.withColumn(c, F.when(F.col(c).isin(7, 9), None).otherwise(F.col(c)))

for c in ["PHYSHLTH", "MENTHLTH"]:
    df = df.withColumn(c, F.when(F.col(c).isin(77, 99), None)
                           .when(F.col(c) == 88, 0)
                           .otherwise(F.col(c)))

df = df.withColumn("INCOME3", F.when(F.col("INCOME3").isin(77, 99), None).otherwise(F.col("INCOME3")))
df = df.withColumn("_BMI5", F.when(F.col("_BMI5") == 9999, None).otherwise(F.col("_BMI5")))

df = df.filter(F.col("_TOTINDA").isNotNull())
print(f"After dropping _TOTINDA=9/null: {df.count()}")

income_median = df.approxQuantile("INCOME3", [0.5], 0.01)[0]
bmi_median = df.approxQuantile("_BMI5", [0.5], 0.01)[0]
smoke_mode = df.groupBy("SMOKE100").count().orderBy(F.desc("count")).first()["SMOKE100"]

df = df.fillna({"INCOME3": income_median, "_BMI5": bmi_median, "SMOKE100": smoke_mode})

remaining_cols = [c for c in cols_needed if c not in ["INCOME3", "_BMI5", "SMOKE100", "_TOTINDA"]]
df_final = df.dropna(subset=remaining_cols)

n_final = df_final.count()
print(f"Final analytic sample: {n_final} (paper reports 422,756)")

unweighted_rate = df_final.filter(F.col("_TOTINDA") == 2).count() / n_final * 100
print(f"Unweighted inactivity rate on cleaned sample: {unweighted_rate:.2f}% (paper reports 22.5%)")
```

```python
# ---------------------------------------------------------------------------
# SILVER LAYER — BRFSS 2024 Cleaned, Typed, Documented
# ---------------------------------------------------------------------------
from pyspark.sql import functions as F

bronze_df = spark.table("workspace.default.bronze_brfss_2024_raw")

def safe_int(col):
    """Cast to int, treating empty/blank strings as null instead of erroring."""
    return F.when((F.col(col) == "") | F.col(col).isNull(), None).otherwise(F.col(col).cast("int"))

SILVER_COLUMNS = {
    "identifiers":   ["_STATE", "SEQNO"],
    "survey_design": ["_LLCPWT", "_PSU", "_STSTR"],
    "outcome":       ["_TOTINDA"],
    "health":        ["GENHLTH", "PHYSHLTH", "MENTHLTH", "_BMI5", "SMOKE100"],
    "socioeconomic": ["EDUCA", "INCOME3", "EMPLOY1"],
    "diseases":      ["DIABETE4", "CVDINFR4", "CVDSTRK3", "HAVARTH4", "ADDEPEV3"],
    "demographic":   ["SEXVAR", "_AGEG5YR", "_RACE"],
}
all_cols = [c for group in SILVER_COLUMNS.values() for c in group]

df = bronze_df.select(*all_cols)
for c in all_cols:
    df = df.withColumn(c, safe_int(c) if c not in ("_STATE", "SEQNO", "_LLCPWT", "_PSU", "_STSTR")
                        else F.when(F.col(c) == "", None).otherwise(F.col(c)))
df = df.withColumn("_LLCPWT", F.col("_LLCPWT").cast("double"))

RECODE_LOG = []

def recode_missing(df, col, missing_codes, zero_code=None, label=""):
    before_nulls = df.filter(F.col(col).isNull()).count()
    if zero_code is not None:
        df = df.withColumn(col, F.when(F.col(col) == zero_code, 0).otherwise(F.col(col)))
    df = df.withColumn(col, F.when(F.col(col).isin(missing_codes), None).otherwise(F.col(col)))
    after_nulls = df.filter(F.col(col).isNull()).count()
    RECODE_LOG.append({
        "column": col, "label": label, "missing_codes": missing_codes,
        "zero_code": zero_code, "nulls_before": before_nulls, "nulls_after": after_nulls,
    })
    return df

df = recode_missing(df, "_TOTINDA", [9], label="outcome: 9=refused")
df = recode_missing(df, "GENHLTH", [7, 9], label="7=don't know, 9=refused")
df = recode_missing(df, "PHYSHLTH", [77, 99], zero_code=88, label="88=none -> 0; 77/99=DK/refused")
df = recode_missing(df, "MENTHLTH", [77, 99], zero_code=88, label="88=none -> 0; 77/99=DK/refused")
df = recode_missing(df, "_BMI5", [9999], label="9999=missing")
df = recode_missing(df, "SMOKE100", [7, 9], label="7=DK, 9=refused")
df = recode_missing(df, "EDUCA", [9], label="9=refused")
df = recode_missing(df, "INCOME3", [77, 99], label="77=DK, 99=refused")
df = recode_missing(df, "EMPLOY1", [9], label="9=refused")
for c in ["DIABETE4", "CVDINFR4", "CVDSTRK3", "HAVARTH4", "ADDEPEV3"]:
    df = recode_missing(df, c, [7, 9], label="7=DK, 9=refused (disease indicator)")
df = recode_missing(df, "_RACE", [9], label="9=DK/refused")

print("Recode log:")
for entry in RECODE_LOG:
    print(f"  {entry['column']:10s} nulls {entry['nulls_before']:>7} -> {entry['nulls_after']:>7}  ({entry['label']})")

n_before_outcome_drop = df.count()
df = df.filter(F.col("_TOTINDA").isNotNull())
n_after_outcome_drop = df.count()
print(f"\nDropped {n_before_outcome_drop - n_after_outcome_drop} rows missing _TOTINDA "
      f"({n_before_outcome_drop} -> {n_after_outcome_drop})")

impute_stats = {}
for c, method in [("INCOME3", "median"), ("_BMI5", "median"), ("SMOKE100", "mode")]:
    missing_pct = df.filter(F.col(c).isNull()).count() / df.count() * 100
    if method == "median":
        value = df.approxQuantile(c, [0.5], 0.01)[0]
    else:
        value = df.groupBy(c).count().orderBy(F.desc("count")).first()[c]
    impute_stats[c] = {"method": method, "value": value, "missing_pct": round(missing_pct, 2)}
    print(f"Imputing {c}: {method}={value}, was {missing_pct:.2f}% missing")

df = df.fillna({c: stats["value"] for c, stats in impute_stats.items()})

low_missing_cols = [c for c in all_cols
                     if c not in list(impute_stats.keys()) + ["_TOTINDA", "_STATE", "SEQNO", "_LLCPWT", "_PSU", "_STSTR"]]
n_before_listwise = df.count()
df_silver = df.dropna(subset=low_missing_cols)
n_after_listwise = df_silver.count()
print(f"\nListwise deletion on {low_missing_cols}: {n_before_listwise} -> {n_after_listwise}")

final_n = df_silver.count()
unweighted_rate = df_silver.filter(F.col("_TOTINDA") == 2).count() / final_n * 100
weighted_rate = df_silver.agg(
    (F.sum(F.when(F.col("_TOTINDA") == 2, F.col("_LLCPWT")).otherwise(0)) /
     F.sum("_LLCPWT") * 100)
).first()[0]

print(f"\n--- Silver layer validation ---")
print(f"Final row count: {final_n} (manuscript: 422,756; prior validation run: 428,992)")
print(f"Unweighted inactivity rate: {unweighted_rate:.2f}% (manuscript: 22.5%)")
print(f"Weighted inactivity rate:   {weighted_rate:.2f}%")

assert final_n > 400000, "Sample size dropped further than expected -- investigate before writing"
assert 20.0 < unweighted_rate < 25.0, "Rate outside plausible range -- investigate before writing"

df_silver.write.format("delta").mode("overwrite").saveAsTable("workspace.default.silver_brfss_2024_clean")

lineage_note = (
    f"Silver layer built from workspace.default.bronze_brfss_2024_raw. "
    f"Special-code-to-null recoding per 2024 BRFSS codebook (see RECODE_LOG in build notebook). "
    f"Imputed INCOME3/_BMI5 (median), SMOKE100 (mode). Listwise deletion on remaining fields. "
    f"Final n={final_n}. Validated: unweighted rate {unweighted_rate:.2f}% vs. manuscript's 22.5% "
    f"(matched analysis, same cleaning logic); weighted rate {weighted_rate:.2f}%. "
    f"Note: _AIDTST4 absent from source file vs. CDC's documented layout (see Bronze notes)."
)
spark.sql(f"COMMENT ON TABLE workspace.default.silver_brfss_2024_clean IS '{lineage_note}'")

print("\nWritten to workspace.default.silver_brfss_2024_clean")
print(f"Table comment set with full transformation lineage.")
```
